```python
import joblib
import pandas as pd
import numpy as np

X_train = joblib.load("../data/processed/X_train.joblib")
X_test = joblib.load("../data/processed/X_test.joblib")

y_train = joblib.load("../data/processed/y_train.joblib")
y_test = joblib.load("../data/processed/y_test.joblib")

print("Entrenamiento:", X_train.shape)
print("Prueba:", X_test.shape)
```

### Desision Tree Classifier

```python
from sklearn.tree import DecisionTreeClassifier

clf = DecisionTreeClassifier(
    max_depth=3,
    random_state=42
)
```

```python
from sklearn.metrics import classification_report

# Entrenar
clf.fit(X_train, y_train)

# Predicciones
train_pred = clf.predict(X_train)
test_pred = clf.predict(X_test)

# Evaluación
print("Train")
print(classification_report(y_train, train_pred))

print("--------------------------------")

print("Test")
print(classification_report(y_test, test_pred))
```

#### Optimizacion 

```python
from sklearn.model_selection import GridSearchCV

# Pruebo distintos valores de parametros para buscar el mejor arbol
param_grid = {
    "max_depth": [3, 5, 7, 10],
    "min_samples_split": [2, 10, 20],
    "min_samples_leaf": [1, 5, 10]
}

grid = GridSearchCV(
    estimator=DecisionTreeClassifier(random_state=42),
    param_grid=param_grid,
    scoring="f1",
    cv=5,
    n_jobs=-1
)

grid.fit(X_train, y_train)
```

```python
print("Mejores parámetros:")
print(grid.best_params_)

print("\nMejor F1 promedio:")
print(round(grid.best_score_, 4))
```

```python
# Tomo el mejor arbol y lo guardo en una variable para usarlo después

mejor_arbol = grid.best_estimator_
```

```python

train_pred = mejor_arbol.predict(X_train)
test_pred = mejor_arbol.predict(X_test)

print("Train")
print(classification_report(y_train, train_pred))

print("--------------------------------")

print("Test")
print(classification_report(y_test, test_pred))
```

Se optimizó el modelo DecisionTreeClassifier utilizando GridSearchCV obteniendo como mejores hiperparámetros max_depth=5, min_samples_leaf=10 y min_samples_split=2.

El modelo optimizado obtuvo un accuracy de 0,75 en Train y 0,74 en Test.

Para la clase fumadores (smoking = 1), el F1-Score aumentó de 0,65 a 0,68 en el conjunto de prueba, mientras que el recall mejoró de 0,69 a 0,76.

La diferencia entre el F1-Score de entrenamiento (0,69) y prueba (0,68) es chica por lo que no se observa señales de sobreajuste

Los resultados muestran una mejora respecto al árbol inicial en la identiifcacion de fumadores

### Random Forest

```python
from sklearn.ensemble import RandomForestClassifier

rf = RandomForestClassifier(
    n_estimators=100,
    random_state=42,
    n_jobs=-1
)

rf.fit(X_train, y_train)

train_pred_rf = rf.predict(X_train)
test_pred_rf = rf.predict(X_test)

print("Train")
print(classification_report(y_train, train_pred_rf))

print("--------------------------------")

print("Test")
print(classification_report(y_test, test_pred_rf))
```

En este caso esta "overfitteando"


```python
from sklearn.metrics import classification_report

rf_controlado = RandomForestClassifier(
    n_estimators=100,
    max_depth=10,
    min_samples_split=10,
    min_samples_leaf=5,
    random_state=42,
    n_jobs=-1
)

rf_controlado.fit(X_train, y_train)

train_pred = rf_controlado.predict(X_train)
test_pred = rf_controlado.predict(X_test)

print("Train")
print(classification_report(y_train, train_pred))

print("--------------------------------")

print("Test")
print(classification_report(y_test, test_pred))
```

Voy a usar randomizerSearchCV propuesto por Chatgpt para buscar los mejores parametros. En clase lo usamos para XGBoost pero no para Random Forest y me parece una buena idea comparar el resultado con los parametros "controlados" que son parametros fijados a modo de limitar la complejidad

```python
from sklearn.model_selection import RandomizedSearchCV


param_dist = {
    "n_estimators": [100, 200],
    "max_depth": [5, 10, 15, 20, None],
    "min_samples_split": [2, 5, 10],
    "min_samples_leaf": [1, 2, 5, 10]
}

random_search = RandomizedSearchCV(
    estimator=RandomForestClassifier(
        random_state=42,
        n_jobs=-1
    ),
    param_distributions=param_dist,
    n_iter=20,
    scoring="f1",
    cv=5,
    random_state=42,
    n_jobs=-1
)

random_search.fit(X_train, y_train)

print("Mejores parámetros:")
print(random_search.best_params_)

print("\nMejor F1 promedio:")
print(round(random_search.best_score_, 4))
```

```python
mejor_rf = random_search.best_estimator_

train_pred_rf = mejor_rf.predict(X_train)
test_pred_rf = mejor_rf.predict(X_test)

print("Train")
print(classification_report(y_train, train_pred_rf))

print("--------------------------------")

print("Test")
print(classification_report(y_test, test_pred_rf))
```

Se evaluaron tres configuraciones del modelo RandomForest buscando el mejor rendimiento.

Random Forest inicial: se entrenó un modelo sin restricciones adicionales obteniendo un F1-Score de 1,00 en entrenamiento y 0,72 en prueba. La diferencia entre ambos resultados evidencia que esta "overfitteando"

Random Forest controlado: se limitaron los árboles utilizando max_depth=10, min_samples_split=10 y min_samples_leaf=5. Se obtuvo un F1-Score de 0,75 en entrenamiento y 0,70 en prueba, con un accuracy de 0,76. La diferencia entre ambos conjuntos se redujo considerablemente, mostrando menor sobreajuste, aunque con una pequeña disminución del rendimiento en prueba.

Random Forest optimizado: se utilizó RandomizedSearchCV para buscar una mejor combinación de hiperparámetros. La configuración seleccionada fue n_estimators=200, max_depth=20, min_samples_split=2 y min_samples_leaf=1.

Este modelo obtuvo un F1-Score de 1,00 en entrenamiento y 0,73 en prueba, con un accuracy de 0,79. Si bien alcanzó el mayor rendimiento en prueba, continúa presentando un overfitting importante.

Conclusión: el Random Forest optimizado obtuvo el mejor F1-Score en prueba (0,73), mientras que el Random Forest controlado presentó una menor diferencia entre entrenamiento y prueba (0,05 frente a 0,27).

El modelo controlado tiene menos overfitting, mientras que el optimizado ofrece un mejor rendimiento predictivo observado. Ambos resultados serán considerados al comparar los demás algoritmos, priorizando el F1-Score de la clase fumador.

### XGBoost

```python
from xgboost import XGBClassifier
from sklearn.metrics import classification_report
```

```python
xgb = XGBClassifier(
    n_estimators=300,
    learning_rate=0.02,
    max_depth=6,
    random_state=42
)
```

```python
xgb.fit(X_train, y_train)
```

```python
train_pred_xgb = xgb.predict(X_train)
test_pred_xgb = xgb.predict(X_test)

print("Train")
print(classification_report(y_train, train_pred_xgb))

print("--------------------------------")

print("Test")
print(classification_report(y_test, test_pred_xgb))
```

Se entrenó un modelo XGBClassifier utilizando 300 estimadores una tasa de aprendizaje de 0,02 y una profundidad máxima de 6 como se hizo en clase.

El modelo obtuvo un F1-Score de 0,74 en entrenamiento y 0,70 en prueba para la clase fumador con un accuracy de 0,77 en el conjunto de test.

En comparación con los modelos anteriores XGBoost obtuvo el mismo F1-Score en prueba que el Random Forest controlado (0,70), mientras que el Random Forest optimizado alcanzó un valor superior (0,73), aunque con overfitting.


### Optimización

```python
param_dist = {
    "n_estimators": [50, 100, 200, 500],
    "max_depth": [3, 5, 7],
    "learning_rate": [0.01, 0.05, 0.001],
    "subsample": [0.6, 0.8, 1]
}
xgb_search = RandomizedSearchCV(
    estimator=XGBClassifier(random_state=42),
    param_distributions=param_dist,
    n_iter=15,
    scoring="f1",
    cv=5,
    random_state=42,
    verbose=2,
    n_jobs=-1
)

xgb_search.fit(X_train, y_train)

print("Mejores parámetros:")
print(xgb_search.best_params_)

print("\nMejor F1 promedio:")
print(round(xgb_search.best_score_, 4))
```

nota: En clase usamos como scoring el fbeta_scorer  usando como beta = 2 para darle mas importancia al recall. Para este trabajo usamos el F1 para dar equilibrio entre precision y recall ya que es el parametro a evaluar en el TP

```python
mejor_xgb = xgb_search.best_estimator_

train_pred_xgb_opt = mejor_xgb.predict(X_train)
test_pred_xgb_opt = mejor_xgb.predict(X_test)

print("Train")
print(classification_report(y_train, train_pred_xgb_opt))

print("--------------------------------")

print("Test")
print(classification_report(y_test, test_pred_xgb_opt))
```

La mejor configuración encontrada fue: n_estimators = 200, max_depth = 7, learning_rate = 0.05, subsample = 0.8

La búsqueda obtuvo un F1-Score promedio de 0,6966.

El modelo optimizado alcanzó un F1-Score de 0,80 en entrenamiento y 0,70 en test y un  accuracy de 0,77.

Si bien la optimización aumentó el rendimiento sobre los datos de entrenamiento no produjo una mejora en el F1-Score de test. La diferencia entre Train y Test aumentó de 0,04 a 0,10, lo que sugiere un mayor overfitting.

El modelo XGBoost inicial presentó un comportamiento más equilibrado entre entrenamiento y test  manteniendo el mismo F1-Score en Test que la versión optimizada.

Por este motivo el XGBoost inicial es candidato para la selección final  dado que la optimización no demostró una mejora en su rendimiento predictivo sobre los datos de prueba.

#### Comparación

```python
import pandas as pd

resultados = pd.DataFrame({
    "Modelo": [
        "Decision Tree inicial",
        "Decision Tree optimizado",
        "Random Forest inicial",
        "Random Forest controlado",
        "Random Forest optimizado",
        "XGBoost inicial",
        "XGBoost optimizado"
    ],
    "F1 Train": [0.66, 0.69, 1.00, 0.75, 1.00, 0.74, 0.80],
    "F1 Test": [0.65, 0.68, 0.72, 0.70, 0.73, 0.70, 0.70]
})

resultados["Diferencia"] = (
    resultados["F1 Train"] - resultados["F1 Test"]
).round(2)

resultados.sort_values(
    by="F1 Test",
    ascending=False
)
```

En la comparación de los resultados vemos que  Random Forest optimizado obtuvo el mayor F1-Score en prueba pero su F1 en train es 1.00. Random Forest controlado y XGBoost inicial muestran diferencias menores entre entrenamiento y test aunque con un F1-Score inferior
